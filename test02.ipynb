```python
# Assigning the value 3 to the variable num
num = 3

# Checking if the value of num is greater than 0
if num > 0:
    # If the condition is True, this block will be executed
    print(num, "is a positive number.")

# This statement is outside the if block and will always be executed
print("This is always printed.")
```

```python
# Assigning a value to the variable x
x = 10

# Check if x is equal to 1
if x == 1:
    print('X is equal to 1')

# Check if x is equal to 5
elif x == 5:
    print('X is equal to 5')

# Check if x is equal to 10
elif x == 10:
    print('X is equal to 10')

# If none of the above conditions are True
else:
    print('X =', x)

```
